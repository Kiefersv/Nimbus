# Nimbus Output

Nimbus provides a comprehensive xarray dataset as output. For comprehensive examples on how to use xarray datasets checkout their [documentation](https://docs.xarray.dev/en/stable/). Here we list the most important physical and analytic output.

## Physical Parameters

The most important outputs of Nimbus are the cloud structure. The two main parameters are:
- 'cloud_mmr': The Mass Mixing Ratio (MMR) of each cloud particle species
- 'cloud_radius': The mean radius of the cloud particles

Based on these values, other cloud parameters can be calculated and their optical properties are defined. To study the cloud structure in more detail, these variables are helpfully:
- 'nucleation_rate': Defines where in the atmosphere new cloud particles are produced.
- 'growth_rate': Defines where in the atmosphere cloud particles grow to larger sizes, often creating the main cloud base.
- 'gas_mmr': Shows the depletion of the cloud species from the gas-phase. This is important to know when the effect of clouds on the gas-phase is studied.
- 'saturation_mmr': Defines the theoretical minimum of gas-phase species. At pressures where this variable is smaller than the 'gas_mmr', phase equilibrium is no longer given.

## Analytic Parameters

Nimbus includes information about the numerics of the run which can be used to assess the run. The most helpful ones are:
- 'species': This lists the cloud species in order they apear in all other arrays.
- 'Did the run finish?': If this is true, the run did not finished. In this case, check 'tfailed' to see which timestep was reached before it was aborted.

These outputs are saved as attributes can be accessed as ds.attrs['variable_name']

| 'variable_name'         | units               | Description                                            |
|-------------------------|---------------------|--------------------------------------------------------|
| **Physical Properties** |                     |                                                        |
| 'species'               | -                   | Name and order of the cloud species                    |
| 'mmw'                   | amu                 | Mean molecular weight                                  |
| 'gravity'               | cm/s2               | Planetary gravity                                      |
| 'metalicity'            | (cm3/cm3)/(cm3/cm3) | Metalicity of the planet compared to solar             |
| 'deep_mmr'              | g/g                 | Interior gas-phase MMR                                 |
| 'ignore_as_nucleator'   | -                   | Species which cannot nucleate                          |
|                         |                     |                                                        |
| **Run settings**        |                     |                                                        |
| 'fsed_init'             | -                   | Sedimentation efficency for initial conditions         |
| 'tstart'                | s                   | Start time of the simulation                           |
| 'tend'                  | s                   | End time of the simulation                             |
| 'tsteps'                | -                   | Number of timesteps in output                          |
| 'ode_rtol'              | -                   | Relative tolarance of the ODE solver                   |
| 'ode_atol'              | -                   | Absolute tolarance of the ODE solver                   |
| 'ode_minimum_mmr'       | g/g                 | Minimum MMR for the ODE solver                         |
| 'static_rg'             | bool                | If True, static radius was used                        |
| 'r_ccn'                 | cm                  | Cloud condensation nucleai radius                      |
| 'cs_mol'                | cm2                 | Molecular cross section                                |
| 'eps_k'                 | erg/mol             | Lennard-Jones potential                                |

| **Analytic**            |                     |                                                        |
| 'total_iterations'      | -                   | Total number of iterations performed                   |
| 'tfailed'               | s                   | Time at which the simulation failed ("None" otherwise) |
| 'Did the run finish?'   | bool                | False if the run failed                                |
|                         |                     |                                                        |
| **Re-start parameters** |                     |                                                        |
| 'y_last'                | -                   | Last ODE solver output, used to restart the run        |
| 'settings'              | -                   | Array of settings (see below)                          |
| 'solver_type'           | -                   | Name of ODE solver used                                |

The 'settings' variables contains an array of floats which define the cloud run. While it is mainly used to re-start old runs, you might want to revisit the settings. The values in this array are in order: [tstart, tend, tsteps, ode_rtol, ode_atol, ode_minimum_mmr, r_ccn, cs_mol, eps_k, r1, rho_ccn, rg_fit_deg]
